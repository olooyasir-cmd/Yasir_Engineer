## 1 · Variables are labelled boxes 📦

In the lab you wrote `name = "Shadrack"`. Let's *really* understand what happened.
A **variable** is a name that points to a value. You create one with `=` (read it as "gets", not "equals").

```python
name = "Yasir Abdul-Rahaman"          # a string (text)
age = 24                   # an integer (whole number)
height_m = 1.75            # a float (decimal number)
is_learning_ai = True      # a boolean (True/False)

print(name, age, height_m, is_learning_ai)
```

**Four core types** — and Python can tell you which is which with `type()`:

```python
print(type(name))
print(type(age))
print(type(height_m))
print(type(is_learning_ai))
```

## 3 · f-strings — the professional way to build text

In the lab you glued text with `+`. That works, but pros use **f-strings**: put an `f` before the
quote, then drop variables inside `{curly braces}`. Cleaner, and it handles numbers automatically.

```python
name = "Ama"
apps_built = 0
goal = 3

# The old way (breaks if you forget str() around numbers):
print("Hi " + name + ", you have built " + str(apps_built) + " apps.")

# The f-string way 🔥
print(f"Hi {name}, you have built {apps_built} apps. Only {goal - apps_built} to go!")
```

Notice `{goal - apps_built}` — you can do **math right inside the braces**.

## 4 · Strings have superpowers (methods)

A **method** is an action a value knows how to do on itself, called with a dot: `value.method()`.
You'll use these constantly when cleaning up AI responses later.

```python
message = "  Build Your First AI App  "

print(message.strip())        # remove spaces at the ends
print(message.lower())        # all lowercase
print(message.upper())        # ALL CAPS
print(message.strip().replace("AI", "Artificial Intelligence"))
print(len(message))           # len() = how long is it (a function, not a method)
```

## 5 · `input()` — programs that listen

`print()` talks. `input()` listens: it pauses, waits for the user to type, and hands you what
they typed as a string. This is the seed of every chatbot you'll ever build.

```python
your_name = input("What is your name? ")
your_goal = input("What do you want to build? ")

print(f"\nWelcome, {your_name}! 🚀")
print(f"By Week 5 you will have built and deployed: {your_goal}")
```

## 6 · Numbers from input need converting

`input()` **always** gives you a string — even if the user types `24`. To do math with it,
convert with `int()` or `float()`:

```python
birth_year = int(input("What year were you born? "))   # int() converts "1999" ➜ 1999
age_now = 2026 - birth_year
print(f"You are about {age_now} years old.")
print(f"When you finish this course you'll still be {age_now} — but with 3 AI apps deployed 😎")
```

## 7 · Put it together — a mini profile card

Everything above in one small program. **Run it, then customise it** — this is the pattern of
