## Challenge 6 · Decisions
Write an `if / elif / else` that sets the variable `verdict` based on `score`:
- 80 or above ➜ `"distinction"`
- 50 to 79 ➜ `"pass"`
- below 50 ➜ `"retry"`

It must work for **any** value of `score`, not just 65 (the check tries several!).

```python
def grade(score):
    # ✍️ YOUR CODE BELOW — set verdict with if/elif/else, based on score
    if score >= 80:
        verdict = "distinction"
    elif score >= 50:
        verdict = "pass"
    else:
        verdict = "retry"


    return verdict
```

```python
# ✅ CHECK 6 — do not edit
ok = True
ok &= check("grade(91) is distinction", grade(91) == "distinction")
ok &= check("grade(80) is distinction (edge!)", grade(80) == "distinction", "80 or ABOVE — use >=")
ok &= check("grade(65) is pass", grade(65) == "pass")
ok &= check("grade(50) is pass (edge!)", grade(50) == "pass", "50 counts as a pass — use >=")
ok &= check("grade(49) is retry", grade(49) == "retry")
record("Challenge 6", ok)
```

## Challenge 8 · Boss level 🐉 — conversation counter
`chat` below is a conversation like the ones your Week 3 bot will store.
Count how many messages have the role `"user"` and store it in `user_count`.
(Loop through `chat`, check each message's `"role"`.)

```python
chat = [
    {"role": "system", "content": "You are helpful."},
    {"role": "user", "content": "Hi!"},
    {"role": "assistant", "content": "Hello!"},
    {"role": "user", "content": "What's 2+2?"},
    {"role": "assistant", "content": "4."},
    {"role": "user", "content": "Thanks!"},
]

user_count = 0

# ✍️ YOUR CODE BELOW
for message in chat:
    if message["role"] == "user":
        user_count += 1

print(f"My counter is: {user_count}")
```

```python
# ✅ CHECK 8 — do not edit
record("Challenge 8", check("user_count is correct", user_count == 3,
      'Loop: for m in chat: ... if m["role"] == "user": user_count += 1'))
```
